```python
# Criando uma lista
lista = [10, 20, 30]

# 1. append()
# Adiciona um elemento no final da lista
lista.append(40)
print(lista)


# 2. insert()
# Insere um elemento em uma posição específica
lista.insert(1, 15)
print(lista)


# 3. remove()
# Remove a primeira ocorrência de um valor
lista.remove(20)
print(lista)


# 4. pop()
# Remove e retorna um elemento
x = lista.pop()
print(x)
print(lista)


# 5. count()
# Conta quantas vezes um valor aparece
print(lista.count(10))


# 6. index()
# Retorna o índice da primeira ocorrência
print(lista.index(30))


# 7. sort()
# Ordena a lista em ordem crescente
lista.sort()
print(lista)


# 8. reverse()
# Inverte a ordem dos elementos
lista.reverse()
print(lista)
```

```python
# declaracao do conjunto (chaves)
frutas = {"maçã", "uva", "maçã"}

# duplicatas sao eliminadas
print(frutas)  # {'maçã', 'uva'}

# teste de pertinencia (acesso)
print("uva" in frutas)  # True

# percorrendo o conjunto
for f in frutas:
    print(f)
```

```python
# lista original de valores
lista = [10, 5, 20, 3, 42, 7]
# filtragem com list comprehension
maiores = [x for x in lista if x > 10]
print(maiores) # [20, 42]
# mesma filtragem com filter()
pares = list(filter(lambda x: x % 2 == 0, lista))
print(pares) # [10, 20, 42]
```

```python
# lista fora de ordem
lista = [10, 5, 20, 3]
# sort() ordena a propria lista
lista.sort()
print(lista) # [3, 5, 10, 20]
# ordem decrescente
lista.sort(reverse=True)
print(lista) # [20, 10, 5, 3]
# sorted() devolve uma nova lista
nova = sorted(lista)
```

```python
# dois conjuntos de valores
A = {1, 2, 3}
B = {3, 4, 5}
# uniao: tudo de A e de B
print(A | B) # {1, 2, 3, 4, 5}
# intersecao: o que ha nos dois
print(A & B) # {3}
# diferenca: so o que e
```

```python
# arquivo: adicionar.py

def adicionar(estoque, produto):
    # evita duplicidade no estoque
    if produto not in estoque:
        estoque.append(produto)  # insere
    return estoque

# teste rapido do modulo
if __name__ == "__main__":
    print(adicionar(["mouse"], "teclado"))
```

```python
# arquivo: remover.py

def remover(estoque, produto):
    # so remove se o item existir
    if produto in estoque:
        estoque.remove(produto)
    else:
        print("Produto nao encontrado")
    return estoque

if __name__ == "__main__":
    print(remover(["mouse", "hd"], "hd"))
```

```python
# arquivo: atualizar.py

def atualizar(estoque, antigo, novo):
    # localiza o indice do produto
    if antigo in estoque:
        i = estoque.index(antigo)
        estoque[i] = novo  # substitui
    return estoque

if __name__ == "__main__":
    print(atualizar(["mose"], "mose", "mouse"))
```

```python
# arquivo: iterar.py

def iterar(estoque):
    # enumerate da indice e valor
    for i, produto in enumerate(estoque, start=1):
        print(i, "-", produto)
    print("Total:", len(estoque))

if __name__ == "__main__":
    iterar(["mouse", "teclado"])
```

```python
# arquivo: pertencimento.py

def existe(estoque, produto):
    # o operador in devolve True/False
    return produto in estoque

if __name__ == "__main__":
    lista = ["mouse", "teclado"]
    if existe(lista, "mouse"):
        print("Ja cadastrado")
    else:
        print("Pode comprar")
```

```python
# --- FUNÇÕES DOS MÓDULOS ---

def adicionar(estoque, produto):
    if produto not in estoque:
        estoque.append(produto)
    return estoque

def remover(estoque, produto):
    if produto in estoque:
        estoque.remove(produto)
    else:
        print("Produto nao encontrado")
    return estoque

def atualizar(estoque, antigo, novo):
    if antigo in estoque:
        i = estoque.index(antigo)
        estoque[i] = novo
    return estoque

def iterar(estoque):
    for i, produto in enumerate(estoque, start=1):
        print(i, "-", produto)
    print("Total:", len(estoque))

def existe(estoque, produto):
    return produto in estoque


# --- CÓDIGO PRINCIPAL ---

estoque = ["mouse", "mose", "hd"]

adicionar(estoque, "teclado")
atualizar(estoque, "mose", "mouse2")
remover(estoque, "hd")

if existe(estoque, "teclado"):
    iterar(estoque)  # relatorio final
```

```python
# Exercicio 1: criar lista de numeros
numeros = [] # lista vazia
qtd = int(input("Quantos numeros? "))

for i in range(qtd):
    n = int(input("Digite um numero: "))
    numeros.append(n) # insere no fim

print("Lista:", numeros)
print("Tamanho:", len(numeros))
```

```python
# Exercicio 2: filtrar numeros pares
texto = input("Numeros separados por espaco: ")
numeros = [int(x) for x in texto.split()]

# filtra os divisiveis por 2
pares = [n for n in numeros if n % 2 == 0]

print("Digitados:", numeros)
print("Pares:", pares)
```

```python
# Exercicio 3: ordenar a lista
texto = input("Numeros separados por espaco: ")
numeros = [int(x) for x in texto.split()]

numeros.sort()  # ordem crescented
print("Crescente:", numeros)

# copia ordenada ao contrario
print("Decrescente:", sorted(numeros, reverse=True))
```

```python
# Exercicio 4: conjunto sem duplicatas
texto = input("Palavras separadas por espaco: ")
palavras = texto.split()

# set() elimina os repetidos
unicos = set(palavras)

print("Digitadas:", len(palavras))
print("Unicas:", len(unicos))
print(sorted(unicos))
```

```python
# Exercicio 5: tupla de estatisticas
texto = input("Notas separadas por espaco: ")

notas = [float(x) for x in texto.split()]

# tupla e imutavel: protege o resultado
estat = (min(notas), max(notas), sum(notas) / len(notas))

print("Minimo:", estat[0])
print("Maximo:", estat[1])
print("Media: %.2f" % estat[2])
```

```python
# --- Exercício Integrador: Python ---

# 1. Criar lista de nomes
nomes = []
qtd = int(input("Quantos nomes deseja digitar? "))

for i in range(qtd):
    nome = input(f"Digite o {i+1}º nome: ")
    nomes.append(nome)

# 2. Remover duplicatas
nomes_unicos = set(nomes)

# 3. Ordenar alfabeticamente
nomes_ordenados = sorted(nomes_unicos)

# Exibindo o resultado final
print("\nLista final ordenada e sem repetições:")
for nome in nomes_ordenados:
    print("-", nome)
```

```python
# 1. Cadastro (Lista)
produtos = []
for i in range(int(input("Quantidade: "))):
    nome = input("Nome: ").strip()
    preco = float(input("Preço: R$ "))
    cat = input("Categoria: ").strip().capitalize()
    produtos.append({"nome": nome, "preco": preco, "cat": cat})

# Filtragem
limite = float(input("\nValor para filtrar: "))
print("Acima:", [f"{p['nome']}: R${p['preco']:.2f}" for p in produtos if p['preco'] > limite])
print("Abaixo:", [f"{p['nome']}: R${p['preco']:.2f}" for p in produtos if p['preco'] < limite])

# 2. Ordenação e Categorias Únicas (Set)
crescente = sorted(produtos, key=lambda x: x["preco"])
categorias = set(p["cat"] for p in produtos)

print(f"\nCrescente: {[p['nome'] for p in crescente]}")
print(f"Decrescente: {[p['nome'] for p in crescente[::-1]]}")
print(f"Categorias únicas: {categorias}")

# 3. Estatísticas Gerais (Tupla) e Relatório Final
precos = [p["preco"] for p in produtos]
estat = (min(precos), max(precos), sum(precos) / len(precos))

print(f"\n=== RELATÓRIO ===")
print(f"Menor: R$ {estat[0]:.2f} | Maior: R$ {estat[1]:.2f} | Média: R$ {estat[2]:.2f}")
```
